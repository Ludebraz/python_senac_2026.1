# Pesquisa: Formas de Armazenamento de Dados Suportadas pelo Python

## 1. Introdução

Python oferece diferentes formas de armazenar e organizar dados. Algumas estruturas mantêm informações temporariamente na memória durante a execução do programa, enquanto outras técnicas permitem salvar os dados em arquivos ou bancos de dados para recuperá-los posteriormente.

## 2. Estruturas de dados em memória

As estruturas de dados em memória permitem organizar e manipular informações durante a execução de um programa. Por si só, elas não garantem que os dados sejam preservados depois que o programa termina.

### 2.1. `list` (Lista)

**O que é:** uma coleção ordenada e mutável de elementos.

**Características:**
- **Ordenada:** mantém a ordem dos elementos.
- **Mutável:** permite adicionar, remover e modificar itens.
- **Permite duplicatas:** o mesmo valor pode aparecer mais de uma vez.
- **Heterogênea:** pode armazenar diferentes tipos de dados.

**Quando usar:** para manter sequências de elementos, especialmente quando a coleção precisa ser modificada.

```python
frutas = ["maçã", "banana", "laranja"]
frutas.append("uva")
frutas.remove("banana")
print(frutas)
```

### 2.2. `tuple` (Tupla)

**O que é:** uma coleção ordenada e **imutável** de itens.

**Características:**
- Mantém a ordem dos elementos.
- Não permite adicionar, remover ou alterar diretamente seus elementos.
- Permite valores duplicados.
- Pode armazenar diferentes tipos de dados.

**Quando usar:** para dados que não devem mudar, para retornar múltiplos valores de uma função e como chave de dicionário quando todos os elementos forem *hashable*.

```python
coordenadas = (-22.505, -43.178)
print(coordenadas[0])
print(coordenadas[1])
```

### 2.3. `dict` (Dicionário)

**O que é:** uma coleção de pares **chave: valor**. No Python 3.7 ou superior, os dicionários preservam a ordem de inserção.

**Características:**
- Cada valor é acessado por uma chave.
- É mutável: permite adicionar, remover e modificar pares.
- As chaves são únicas e precisam ser *hashable*.
- Os valores podem ser de diferentes tipos.

**Quando usar:** para representar registros, organizar dados semelhantes a JSON, pesquisar por chave e contar ocorrências.

```python
produto = {"nome": "Teclado", "preco": 89.90, "estoque": 15}
print(produto["nome"])
produto["preco"] = 94.90
produto["marca"] = "Exemplo"
print(produto)
```

### 2.4. `set` (Conjunto)

**O que é:** uma coleção sem ordem garantida de elementos únicos.

**Características:**
- Não se deve depender da ordem de exibição.
- Elimina valores duplicados.
- É mutável: permite adicionar e remover elementos.
- Seus elementos precisam ser *hashable*; listas e dicionários não podem ser adicionados diretamente.

**Quando usar:** para eliminar duplicatas, realizar união/interseção/diferença e testar rapidamente se um elemento pertence ao conjunto.

```python
numeros = {1, 2, 2, 3, 4}
numeros.add(5)
print(numeros)
print(3 in numeros)
```

Exemplo de operações entre conjuntos:

```python
grupo_a = {1, 2, 3}
grupo_b = {3, 4, 5}
print("União:", grupo_a | grupo_b)
print("Interseção:", grupo_a & grupo_b)
print("Diferença:", grupo_a - grupo_b)
```

### 2.5. Comparação entre as estruturas

| Estrutura | Ordenada | Mutável | Permite duplicatas | Uso principal |
|---|---|---|---|---|
| `list` | Sim | Sim | Sim | Sequências modificáveis |
| `tuple` | Sim | Não | Sim | Dados que não devem mudar |
| `dict` | Preserva ordem de inserção (Python 3.7+) | Sim | Chaves não; valores podem se repetir | Associar chaves a valores |
| `set` | Sem ordem garantida | Sim | Não | Valores únicos e operações de conjuntos |

## 3. Métodos das estruturas de dados em memória

Os métodos são operações disponibilizadas pelos objetos Python para facilitar a manipulação de dados.

### 3.1. Métodos de `list`

| Método | Função |
|---|---|
| `append(x)` | Adiciona um elemento ao final. |
| `extend(iterável)` | Adiciona vários elementos. |
| `insert(i, x)` | Insere em uma posição específica. |
| `remove(x)` | Remove a primeira ocorrência de um valor. |
| `pop(i)` | Remove e retorna um elemento; sem argumento, remove o último. |
| `clear()` | Remove todos os elementos. |
| `index(x)` | Retorna o índice da primeira ocorrência. |
| `count(x)` | Conta as ocorrências de um valor. |
| `sort()` | Ordena a lista. |
| `reverse()` | Inverte a ordem. |
| `copy()` | Cria uma cópia superficial. |

```python
numeros = [5, 2, 8, 2]
numeros.append(10)
numeros.insert(1, 7)
numeros.remove(2)
print("Lista:", numeros)
print("Quantidade de 2:", numeros.count(2))
print("Índice de 8:", numeros.index(8))
numeros.sort()
print("Ordenada:", numeros)
numeros.reverse()
print("Invertida:", numeros)
ultimo = numeros.pop()
print("Removido:", ultimo)
copia = numeros.copy()
numeros.clear()
print("Lista original:", numeros)
print("Cópia:", copia)
```

`sort()` e `reverse()` modificam a lista original e retornam `None`. Para obter uma lista ordenada sem modificar a original, use `sorted(lista)`.

### 3.2. Métodos de `tuple`

A tupla é imutável e possui dois métodos principais:

| Método | Função |
|---|---|
| `count(x)` | Conta quantas vezes um valor aparece. |
| `index(x)` | Retorna o índice da primeira ocorrência. |

```python
numeros = (10, 20, 30, 20, 40)
print("Quantidade de 20:", numeros.count(20))
print("Índice de 30:", numeros.index(30))
```

É possível criar uma nova tupla a partir da original, sem alterar a primeira:

```python
numeros = (10, 20, 30)
nova_tupla = numeros + (40,)
print(nova_tupla)
```

### 3.3. Métodos de `dict`

| Método | Função |
|---|---|
| `get(chave)` | Retorna o valor ou um valor padrão. |
| `keys()` | Retorna uma visão das chaves. |
| `values()` | Retorna uma visão dos valores. |
| `items()` | Retorna uma visão dos pares chave-valor. |
| `update()` | Adiciona ou atualiza pares. |
| `pop(chave)` | Remove uma chave e retorna seu valor. |
| `popitem()` | Remove e retorna o último par inserido. |
| `clear()` | Remove todos os pares. |
| `copy()` | Cria uma cópia superficial. |
| `setdefault(chave, valor)` | Retorna o valor existente ou insere a chave com valor padrão. |
| `fromkeys(iterável, valor)` | Cria um dicionário com as chaves fornecidas. |

```python
produto = {"nome": "Teclado", "preco": 100.00, "estoque": 10}
print("Nome:", produto.get("nome"))
print("Chaves:", produto.keys())
print("Valores:", produto.values())
print("Pares:", produto.items())
produto.update({"preco": 120.00, "marca": "Exemplo"})
print("Atualizado:", produto)
print("Estoque removido:", produto.pop("estoque"))
produto.setdefault("categoria", "Informática")
print("Categoria:", produto)
copia = produto.copy()
print("Cópia:", copia)
print("Último par removido:", produto.popitem())
produto.clear()
print("Dicionário vazio:", produto)
```

Exemplo de `fromkeys()`:

```python
campos = ["nome", "idade", "cidade"]
pessoa = dict.fromkeys(campos, "Não informado")
print(pessoa)
```

`keys()`, `values()` e `items()` retornam visões dinâmicas do dicionário. Se `fromkeys()` receber um valor mutável, como uma lista, esse mesmo objeto será compartilhado entre as chaves.

### 3.4. Métodos de `set`

| Método | Função |
|---|---|
| `add(x)` | Adiciona um elemento. |
| `update(iterável)` | Adiciona elementos de iteráveis. |
| `remove(x)` | Remove um elemento; gera erro se não existir. |
| `discard(x)` | Remove um elemento sem erro se não existir. |
| `pop()` | Remove e retorna um elemento arbitrário. |
| `clear()` | Remove todos os elementos. |
| `union()` | Retorna a união. |
| `intersection()` | Retorna a interseção. |
| `difference()` | Retorna os elementos do primeiro conjunto que não estão nos demais. |
| `symmetric_difference()` | Retorna elementos exclusivos de cada conjunto. |
| `issubset()` | Verifica se é subconjunto. |
| `issuperset()` | Verifica se contém todos os elementos de outro conjunto. |
| `isdisjoint()` | Verifica se não há elementos em comum. |
| `intersection_update()` | Mantém apenas os elementos comuns no conjunto original. |
| `difference_update()` | Remove elementos presentes nos outros conjuntos. |
| `symmetric_difference_update()` | Mantém os elementos exclusivos de cada conjunto. |
| `copy()` | Cria uma cópia superficial. |

```python
grupo_a = {1, 2, 3, 4}
grupo_b = {3, 4, 5, 6}
grupo_a.add(7)
grupo_a.update([8, 9])
print("Após adicionar:", grupo_a)
grupo_a.discard(9)
print("Após descartar 9:", grupo_a)
print("União:", grupo_a.union(grupo_b))
print("Interseção:", grupo_a.intersection(grupo_b))
print("Diferença:", grupo_a.difference(grupo_b))
print("Diferença simétrica:", grupo_a.symmetric_difference(grupo_b))
print("Subconjunto:", {1, 2}.issubset(grupo_a))
print("Superconjunto:", grupo_a.issuperset({1, 2}))
print("Disjuntos:", {10, 11}.isdisjoint(grupo_a))
grupo_a.intersection_update(grupo_b)
print("Após intersection_update:", grupo_a)
grupo_a.clear()
print("Conjunto vazio:", grupo_a)
```

**Observações:**
- `remove(x)` gera `KeyError` se o elemento não existir; `discard(x)` não gera erro nessa situação.
- `pop()` não permite escolher qual elemento será removido.
- Os métodos sem `_update` retornam novos conjuntos; as versões com `_update` modificam o conjunto original.

### 3.5. Comparação dos principais métodos

| Operação | `list` | `tuple` | `dict` | `set` |
|---|---|---|---|---|
| Adicionar elementos | `append()`, `extend()`, `insert()` | Não possui método de inclusão | `update()` ou atribuição por chave | `add()`, `update()` |
| Remover elementos | `remove()`, `pop()`, `clear()` | Não possui método de remoção | `pop()`, `popitem()`, `clear()` | `remove()`, `discard()`, `pop()`, `clear()` |
| Consultar | `count()`, `index()` | `count()`, `index()` | `get()`, `keys()`, `values()`, `items()` | Teste com `in` |
| Ordenar | `sort()` | Não ordena a tupla original | Preserva ordem de inserção | Não garante ordem |
| Criar cópia | `copy()` | Não possui método `copy()` | `copy()` | `copy()` |

## 4. Persistência de dados

O armazenamento em memória, como listas, tuplas e dicionários, é temporário: os dados mantidos apenas na memória do programa normalmente deixam de estar disponíveis quando a execução termina.

Para que as informações sobrevivam ao encerramento do programa e possam ser recuperadas em uma execução futura, utilizamos técnicas de **persistência de dados**.

Em Python, há várias abordagens, desde a manipulação de arquivos até o uso de bancos de dados.

### 4.1. Persistência em arquivos simples (texto e binário)

Python oferece funções nativas para trabalhar com arquivos, como `open()`, além de métodos como `read()` e `write()`.

- **`open()`:** abre um arquivo para leitura, escrita ou outras operações.
- **`read()`:** lê o conteúdo de um arquivo.
- **`write()`:** grava dados em um arquivo.

Arquivos de texto armazenam informações textuais. Arquivos binários armazenam dados em formato de bytes, apropriado para determinados formatos e conteúdos.

### 4.1.1. Arquivos de texto comum (`.txt`)

**O que é:** um arquivo que armazena dados como texto, normalmente strings organizadas em linhas.

**Vantagens:**
- Simples de criar e utilizar.
- Pode ser lido e editado por pessoas em um editor de texto.
- Útil para registros, anotações e configurações simples.

**Desvantagens:**
- É necessário definir como os dados serão organizados e interpretados.
- Valores podem precisar ser convertidos de texto para tipos Python, como inteiros ou números decimais.
- Não oferece, por si só, recursos de consulta e relacionamento como os de um banco de dados.

**Quando usar:** logs, anotações, configurações básicas e dados que precisam ser facilmente lidos ou editados por uma pessoa.

#### Exemplo: gravando informações em um arquivo

```python
with open("anotacoes.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Minha primeira gravação em arquivo.\n")
    arquivo.write("Os dados podem ser lidos depois.\n")
```

O modo `"w"` cria o arquivo se ele não existir e substitui seu conteúdo se ele já existir. `encoding="utf-8"` ajuda a preservar caracteres como acentos.

#### Exemplo: lendo informações de um arquivo

```python
with open("anotacoes.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
```

O modo `"r"` abre o arquivo para leitura. O uso de `with` garante que o arquivo seja fechado corretamente ao final do bloco.

#### Exemplo: acrescentando informações sem apagar o conteúdo anterior

```python
with open("anotacoes.txt", "a", encoding="utf-8") as arquivo:
    arquivo.write("Esta linha foi acrescentada.\n")
```

O modo `"a"` acrescenta conteúdo ao final do arquivo, preservando o que já estava gravado.

## 5. Conclusão parcial

As estruturas `list`, `tuple`, `dict` e `set` permitem organizar dados em memória de maneiras diferentes. A escolha depende de características como ordenação, possibilidade de alteração, duplicatas e forma de acesso aos elementos.

Quando os dados precisam continuar disponíveis depois que o programa termina, é necessário utilizar persistência. Arquivos de texto são uma opção simples para guardar informações; outros métodos, como CSV, JSON e bancos de dados, podem atender a necessidades mais específicas e poderão ser acrescentados às próximas seções da pesquisa.
