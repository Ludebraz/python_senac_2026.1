# 1. Lista (`list`)

## O que é

Uma **lista** é uma coleção **ordenada** e **mutável** de elementos.

Cada elemento possui uma posição chamada de **índice**, que começa em `0`.

# 2. Tupla (`tuple`)

## O que é

Uma **tupla** é uma coleção **ordenada** e **imutável**.

Depois de criada, não é possível adicionar, remover ou alterar diretamente seus elementos.

## Características

| Característica | Significado |
|---|---|
| Ordenada | Mantém a ordem dos elementos. |
| Imutável | Não permite alteração após a criação. |
| Indexada | Permite acesso por posição. |
| Permite duplicatas | Pode conter valores repetidos. |
| Heterogênea | Pode armazenar tipos diferentes. |

## Quando utilizar

- Para dados que não devem ser alterados.
- Para coordenadas, configurações fixas ou registros imutáveis.
- Para retornar vários valores de uma função.

## dict (Dicionário)

O que é: uma coleção de pares chave: valor. No Python 3.7 ou superior, os dicionários preservam a ordem de inserção, embora sua principal finalidade seja associar chaves a valores.

### Características:

Chave-valor: cada valor é acessado por meio de uma chave.
Mutável: permite adicionar, remover e modificar pares.
Chaves únicas: uma chave não pode aparecer mais de uma vez; atribuir um novo valor a uma chave existente substitui o valor anterior.
Chaves hashable: strings, números e tuplas compostas por elementos hashable são exemplos de chaves válidas. Listas e dicionários não podem ser usados como chaves.
Valores variados: os valores podem ser de diferentes tipos, inclusive listas e outros dicionários.

### Quando usar:

Para representar dados estruturados, como registros de pessoas ou produtos.
Para organizar dados semelhantes aos encontrados em arquivos JSON.
Para pesquisar informações usando uma chave.
Para contar ocorrências de elementos.

### Exemplo:

produto = {
    "nome": "Teclado",
    "preco": 89.90,
    "estoque": 15
}

print(produto["nome"])

produto["preco"] = 94.90
produto["marca"] = "Exemplo"

print(produto)

O conjunto mantém apenas uma ocorrência de cada valor. A ordem exibida pode variar.

### Exemplo de operações entre conjuntos:

grupo_a = {1, 2, 3}
grupo_b = {3, 4, 5}

print(grupo_a | grupo_b)  # União

print(grupo_a & grupo_b)  # Interseção

print(grupo_a - grupo_b)  # Diferença
