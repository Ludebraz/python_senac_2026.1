# Criando Dicionários

```python
# Dicionario Vazio
d1={}
d2=dict()

# Dicionários com elementos
telefones = {"Alice": "1234-5678", "Bob": "8765-4321"}
print(telefones)
```

```python
d3= dict(nome= "Carlos", idade= 25)
print(d3)

# Apartir de uma lista de tuplas
listaDeTuplas= [("a", 1), ("b", 2)]
d4=dict(listaDeTuplas)
print(d4)
```

```python
chaves = ["nome", "idade", "cidade"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)

# Resultado: {'nome': 'N/A', 'idade': 'N/A', 'cidade': 'N/A'}
print(usuario)

usuario["nome"]= "Lude Braz"
usuario["idade"]= 32
usuario["cidade"]= "Petrópolis"
usuario["cep"]="25745-400"
print(usuario)
```
