```python
chaves=["nome", "dt_nascimento", "cep", "lagradouro", "bairro", "cidade", "estado"]
valor_padrao= "N/A"
usuario=dict.fromkeys(chaves, valor_padrao)

print(usuario)
```

```python
usuario["nome"]=input("Digite o seo nome :")
```

```python
usuario["dt_nascimento"]=input("Digite sua data de nascimento :")
```

```python
usuario["cep"]=input("Digite seu CEP :")
```

```python
usuario["lagradouro"]=input("Digite o seu lagradouro :")
```

```python
usuario["bairro"]=input("Digite o seu Bairro :")
```

```python
usuario["cidade"]=input("Digite a sua cidade :")
```

```python
usuario["estado"]=input("Digite o seu Estado :")
```

```python
print(usuario)
```

```python
for chave, valor in usuario.items():
    print(f"{chave.capitalize():_<20}: {valor.capitalize()}")
```
