### 1️⃣ Cadastro de produtos
📚 Exercício:
Crie um programa que recebe o nome e o preço de 5 produtos, armazene-os em uma tupla e exiba uma listagem formatada.

```python
produtos = (
    ("Arroz", 5.99, 10),
    ("Feijão", 7.49,  3),
    ("Leite", 4.89, 10,),
    ("Óleo", 9.99, 4),
    ("Açúcar", 3.29, 5)
)
valorTotal= 0
valorProduto=0

print("LISTA DE PRODUTOS")
for nome, preco, qtda in produtos:
    valorProduto=preco*qtda
    #valorTotal=valorTotal+valorProduto
    valorTotal+=valorProduto
    print(f"{nome:.<20} R$ {preco:.2f} QT {qtda:.2f} Valor Total {valorProduto:.2f}" )

print(f"Valor Total de Compra: R${valorTotal}")
```

```python
import random

idades=[]
for _ in range(100):
    idades.append(random.randint(1,92))

maiores = 0
for idade in idades:
    if idade >= 18:
        maiores += 1

print(f"Quantidade de maiores de idade: {maiores}")
```

```python
import random
# Gerando os 5 lançamentos com for tradicional
lancamentosLista=[]
for _ in range(5):
    lancamentosLista.append(random.randint(1,6))

lancamentos= tuple(lancamentosLista)
print(f"Rsultados {lancamentos}")

#Contagem tradicional
for num in range(1,7):
    print(f"{num} saiu {lancamentos.count(num)} vezes")
```
