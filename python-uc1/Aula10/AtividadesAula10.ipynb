```python
import random
def criar_matriz(qt_linha=4, qt_colunas=4):
    matriz=[]
    for linha in range(qt_linha):
        linha=[]
        for _coluna in range(qt_colunas):
            linha.append(random.randint(1,100))
        matriz.append(linha)
    return (matriz)
matriz=criar_matriz(5,5)
for linha in matriz:
    print(linha)
```

```python
# Inicializa uma matriz 4x4 vazia
matriz = []

print("Digite os valores da matriz 4x4:")

# Loop para ler as 4 linhas
for i in range(4):
    linha = []  # Inicializa uma linha vazia
    
    # Loop para ler os 4 elementos de cada linha
    for j in range(4):
        # valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
        linha.append(random.randint(1,100))  # Adiciona o valor à linha
    
    matriz.append(linha)  # Adiciona a linha à matriz

# Exibe a matriz formatada
print("\nMatriz 4x4:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")  # Formata cada elemento com 4 espaços
    print()  # Pula linha após cada linha da matriz
```

```python
def imprime_matriz(matriz=[]):
    for linha in matriz:        
        for elemento in linha :
            print(f"{elemento:4d}",end=" ")
        print()
```

```python
minha_matriz=criar_matriz(5,5)
```

```python
imprime_matriz(minha_matriz)
```
