### 4️⃣ Jogo de Adivinhação
O usuário precisa adivinhar um número entre 1 e 10. Ele tem 3 tentativas!

```python
numero_secreto = 7
tentativa1= int(input("Tente adivinha o numero(1 a 10):"))
tentativa2= int(input("tente novamente"))
tentativa3= int(input("tente novament"))
if tentativa1 or tentativa2 or tentativa3 == numero_secreto:
    print("Parabéns!Você acertou!!")
else:
    print("Que pena! O numero era", numero_secreto )
```

```python
import random
segredo=random.randint(1,10)
for i in range(3):
    numero=int(input("Digite um número : "))
    if numero == segredo :
        print("Parabéns! Você acertou!")
        break
    else:
        print("Você errou, tente novamente")
print(f"O numero secreto era .:{segredo}:.")
```

```python
import random
segredo=random.randint(0,10)
tentativas=0
while tentativas < 3:
    numero=int(input("Digite um número : "))
    if numero == segredo :
        print("Parabéns! Você acertou!")
        break
    else:
        print("Você errou, tente novamente")
    tentativas = tentativas+1
print(f"O numero secreto era .:{segredo}:.")
```
