# Orientação objeto

```python
# Definindo uma classe chamada 'Carro'
class Carro:
    # Método construtor (__init__) para inicializar os atributos
    def __init__(self, marca, modelo, cor,  ano="2026"):
        self.marca = marca      # Atributo
        self.modelo = modelo    # Atributo
        self.ano = ano          # Atributo
        self.cor= cor
        self.potencia= 1.0
        self.cambio="auto"
        self.qt_portas= "4"
        self.dir_hidraulica=True
        self.tp_tracao=4
        self._cambio_automatico=False
        self._4p4=False
        self.__ligado=False
        self.tipodecombustivel="Hibrido"
    
    # Método para ligar o carro
    def ligar(self):
        if not self.__ligado:
            self.__ligado = True
            print(f"{self.marca} {self.modelo} está ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está ligado.")

    # Método para desligar o carro
    def desligar(self):
        if self.__ligado:
            self.__ligado = False
            print(f"{self.marca} {self.modelo} está desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está desligado.")
    
    def mudarCor(self, cor):
        self.cor=cor
        return True

    def mudar_cambio(self, cambio):
        cambio=cambio.lower()
        if cambio== "auto":
            self.cambio.tp_cambio="auto"
        elif cambio=="manual":
            self.tp_cambio="manual"
        else:
            return False
        return True

    def mudar_tracao(self):
        if self.tp_tracao==2 :
            self.tp_tracao=4
        else:
            self.tp_tracao=2
        return True

    def mudar_potencia(self, pot="1.0"):
        try: 
            vl_pot=float(pot)
            self.potencia= vl_pot
            return True
        except:
            return False

    
    # Método para exibir informações do carro
    def exibir_info(self):
        status = "ligado" if self.__ligado else "desligado"
        print(f"{self.marca} {self.modelo} ({self.ano}) está {status}.")
```

```python
meu_carro=Carro("VW","Gol","Cinza", 2013)
meuCarro2=Carro("Fiat", "Palio", "Branco")
meuCarro3=Carro("Toyota", "CorolaCross", "Azul" , "2024")
```
