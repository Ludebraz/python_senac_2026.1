# 8: Função contar_vogais

## Contar o número de vogais em uma string.

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador
    
print("Número de vogais em 'Python':", contar_vogais("Camila"))
```

# 2 Crie um programa Python que use a função main() para calcular e imprimir a soma de dois números fornecidos pelo usuário.

```python
def main():
   num1 = float(input("Digite o primeiro número: "))
   num2 = float(input("Digite o segundo número: "))
   soma = num1 + num2
   print(f"A soma é: {soma}")

if __name__ == "__main__":
   main()
```
